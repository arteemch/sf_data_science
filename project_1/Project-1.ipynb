3. Следующим этапом преобразуем признак **"Опыт работы"**. Его текущий формат - это: **<Опыт работы: n лет m месяцев, периоды работы в различных компаниях…>**. 

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…
    * Опыт работы 4 года…
    * Опыт работы 11 месяцев…
    * Учитывайте эту особенность в вашем коде

Учитывайте эту особенность в вашем коде

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
def experience_converting(text):
    if not isinstance(text,str) or text=='Не указано': # Учитываем пропуски
        return None 
    text=text.split()[:6] # Убираем лишнюю информацию

    if 'год' in text or 'года' in text or 'лет' in text:
        if 'месяц' in text or 'месяца' in text or 'месяцев' in text:
            return int(text[2])*12 + int(text[4]) # Возвращаем кол-во лет*12 + кол-во месяцев
        else:
            return int(text[2])*12 # Возвращаем кол-во лет*12
    else:
        return int(text[2]) # Возвращаем кол-во месяцев


hh_data3['Опыт работы (месяц)'] = hh_data3['Опыт работы'].apply(experience_converting)
hh_data4 = hh_data3.drop('Опыт работы', axis=1) # Создал новый датафрейм, чтобы не возникало ошибок при преобразовании оригинального
display(hh_data4['Опыт работы (месяц)'].head())

# Контрольный вопрос: чему равен медианный опыт работы (в месяцах) в нашей таблице?
print(f'Медианный опыт работы в месяцах: {hh_data4['Опыт работы (месяц)'].median()}')
# Ответ: 100
```

*Совет: обратите внимание на то, что структура текста может меняться в зависимости от указания ближайшего метро. Учите это, если будете использовать порядок слов в своей программе.*


```python
million_cities = ['Новосибирск', 'Екатеринбург', 'Нижний Новгород', 'Казань', 'Челябинск', 'Омск', 'Самара', 'Ростов-на-Дону', 'Уфа', 'Красноярск', 'Пермь', 'Воронеж', 'Волгоград']

def cities_convertion(text): # Разбиваем города на категории
    text=text.split(' , ')
    if text[0] == 'Москва':
        return 'Москва'
    elif text[0] == 'Санкт-Петербург':
        return 'Санкт-Петербург' 
    elif text[0] in million_cities:
        return 'город-миллионник'
    else:
        return 'другие'

hh_data4['Город'] = hh_data4['Город, переезд, командировки'].apply(cities_convertion)



def moving_convertion(text): # Разбиваем на готовность к переезду
    text=text.split(' , ')
    if 'не готов к переезду' in text or 'не готова к переезду' in text:
        return False
    else:
        return True

hh_data4['Готовность к переезду'] = hh_data4['Город, переезд, командировки'].apply(moving_convertion)



def worktrips_convertion(text): #Разбиваем на готовность к командировкам
    text=text.split(' , ')
    if 'не готов к командировкам' in text or 'не готова к командировкам' in text:
      return False
    else:
        return True

hh_data4['Готовность к командировкам'] = hh_data4['Город, переезд, командировки'].apply(worktrips_convertion)

hh_data5 = hh_data4.drop('Город, переезд, командировки', axis=1) # Создал новый датафрейм, чтобы не возникало ошибок при преобразовании оригинального
display(hh_data5[['Город','Готовность к переезду','Готовность к командировкам']].head())


# Контрольные вопросы:
# Сколько процентов соискателей живут в Санкт-Петербурге? Ответ округлите до целого.
display(round(hh_data5['Город'].value_counts(normalize=True)*100))
# Ответ: 11

# Сколько процентов соискателей готовы одновременно и к переездам, и к командировкам? Ответ округлите до целого.
display(round(hh_data5[['Готовность к переезду', 'Готовность к командировкам']].value_counts(normalize=True)*100))
# Ответ: 32 
```

Присоединить к таблице с резюме таблицу с курсами по столбцам с датой и названием валюты (подумайте, какой тип объединения надо выбрать, чтобы в таблице с резюме сохранились данные о заработной плате, изначально представленной в рублях). Значение close для рубля заполнить единицей 1 (курс рубля самого к себе)
4. Умножить сумму желаемой заработной платы на присоединенный курс валюты (close) и разделить на пропорцию (обратите внимание на пропуски после объединения в этих столбцах), результат занести в новый столбец "ЗП (руб)".


```python
# Перевести признак "Обновление резюме" из таблицы с резюме в формат datetime и достать из него дату. 
# В тот же формат привести признак "date" из таблицы с валютами.
currencies_data = pd.read_csv('C:\PYTHON\IDE\PROJECT-1\data\ExchangeRates.csv', sep=',')
currencies_data['date'] = pd.to_datetime(currencies_data['date'], format='mixed')
hh_data7['Обновление резюме'] = pd.to_datetime(hh_data7['Обновление резюме'], format='mixed').dt.normalize() #убираем время


# Выделить из столбца «ЗП» сумму желаемой заработной платы и наименование валюты, в которой она исчисляется. 
# Наименование валюты перевести в стандарт ISO.
def extract_salary(salary):
    return int(salary.split()[0])

def extract_currency(salary):
    salary = salary.split()[1]
    if not isinstance(salary, str):
        return None
    if salary == 'грн.':
        salary = 'UAH'
    elif salary == 'бел.руб.':
        salary = 'BYN'
    elif salary == 'сум':
        salary = 'UZS'
    else:
        pass
    return salary

hh_data7['ЗП (без валюты)'] = hh_data7['ЗП'].apply(extract_salary)

hh_data7['Валюта'] = hh_data7['ЗП'].apply(extract_currency)


# Присоединить к таблице с резюме таблицу с курсами по столбцам с датой и названием валюты.
# Значение close для рубля заполнить единицей 1 (курс рубля самого к себе).
hh_data8 = hh_data7.merge(
    currencies_data,
    left_on=['Обновление резюме', 'Валюта'], # Объединяем таблицы по соответствию дат и валют 
    right_on=['date','currency'],
    how = 'left'
)

hh_data8.loc[hh_data8['Валюта']=='руб.', 'close'] = 1 # Заполняем close единицами для рубля


def convert_to_rub(data):
    if data['Валюта'] == 'руб.':
        return data['ЗП (без валюты)'] # Если рубли, то оставляем ЗП как есть
    else:
        return data['close'] / data['proportion'] * data['ЗП (без валюты)']


hh_data8['ЗП (руб)'] = hh_data8.apply(convert_to_rub, axis=1) # Применяем функцию конвертации к датафрейму и добавляем новый признак


#Контрольный вопрос: Чему равна желаемая медианная заработная плата соискателей в нашей таблице (в рублях)? 
# Ответ приведите в тысячах рублей и округлите до целого.
display(f'Медианная ЗП в рублях: {round(hh_data8['ЗП (руб)'].median()/1000)} тысяч.')
#ответ: 59 тыс.


# Удаляем лишние столбцы:
cols_to_delete = ['ЗП', 'ЗП (без валюты)', 'Валюта', 'currency', 'per', 'date', 'time', 'close', 'vol', 'proportion']
hh_data9 = hh_data8.drop(cols_to_delete, axis=1)
display(hh_data9.head())


                         
```

Попробуйте построить распределение признака в **логарифмическом масштабе**. Добавьте к графику линии, отображающие **среднее и границы интервала метода трех сигм**. Напомним, сделать это можно с помощью метода axvline. Например, для построение линии среднего будет иметь вид:

В какую сторону асимметрично логарифмическое распределение? Напишите об этом в комментарии к графику.
Найдите выбросы с помощью метода z-отклонения и удалите их из данных, используйте логарифмический масштаб. Давайте сделаем послабление на **1 сигму** (возьмите 4 сигмы) в **правую сторону**.

Выведите таблицу с полученными выбросами и оцените, с каким возрастом соискатели попадают под категорию выбросов?

```python
fig, ax = plt.subplots(1, 1, figsize=(15, 6))
log_age = np.log(hh_data12['Возраст'] + 1)
histplot = sns.histplot(log_age, bins=30, ax=ax)
histplot.axvline(log_age.mean(), color='k', lw=2)
histplot.axvline(log_age.mean()+ 3 * log_age.std(), color='k', ls='--', lw=2)
histplot.axvline(log_age.mean()- 3 * log_age.std(), color='k', ls='--', lw=2)
histplot.set_title('Распределение возраста соискателей')
histplot.set_xlabel('Возраст, лет (логарифмическая шкала)')
histplot.set_ylabel('Число соискателей')
plt.show()

# Найдите выбросы с помощью метода z-отклонения и удалите их из данных, используйте логарифмический масштаб. 
# Давайте сделаем "послабление" на 1 сигму (возьмите 4 сигмы) в правую сторону. 
def outliers_z_score(data, feature):
    x = np.log(data[feature]+1)
    mean = x.mean()
    sigma = x.std()
    lower_bound = mean - 3 * sigma
    upper_bound = mean + 4 * sigma # Делаем послабление на 1 сигму
    outliers = data[(x < lower_bound) | (x > upper_bound)] # Выбросы
    cleaned = data[(x >= lower_bound) & (x <= upper_bound)] # Очищенное от выбросов
    return outliers, cleaned
outliers6, cleaned6 = outliers_z_score(hh_data12, 'Возраст')

# Выведите таблицу с полученными выбросами и оцените, с каким возрастом соискатели попадают под категорию выбросов?
# Сколько выбросов вы нашли с помощью метода z-отклонений?
display(outliers6)
# Ответ: под категорию выбросов попали 3 соискателя в возрасте 15, 15 и 100 лет. 


#BONUS. Финальный датафрейм со всеми преобразованиями и очистками
hh_final_dataframe = cleaned6 
```
